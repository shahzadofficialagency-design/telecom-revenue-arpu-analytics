# 04 — Exploratory Data Analysis
Charts are saved to `screenshots/eda_*.png`; every KPI quoted in `reports/business_insights.md` is computed in `src/run_eda_and_kpis.py` and stored in `reports/_kpi_results.json`. Nothing is hand-typed.


```python
import pandas as pd, matplotlib.pyplot as plt
rev = pd.read_csv('../data/processed/revenue_clean.csv', parse_dates=['month'])
snap = pd.read_csv('../data/processed/subscriber_snapshot_clean.csv', parse_dates=['month'])
monthly = rev.groupby('month').net_revenue.sum()
monthly.plot(marker='o', title='Monthly Net Service Revenue (PKR) — Simulated')
plt.ylabel('PKR'); plt.show()
```

```python
subs = snap.groupby('month').average_active_subscribers.sum()
arpu = (rev.groupby('month').net_revenue.sum() / subs)
arpu.plot(marker='o', color='darkorange', title='Monthly ARPU (PKR)')
plt.ylabel('PKR'); plt.show()
print(arpu.round(2))
```
